```python
import pandas as pd

df = pd.read_csv("/content/analise-missoes-lunares/Moonlanding.csv")

# Atividade 1 - Numero de missões por agencia

missoes_por_agencia = df["Operator"].value_counts()

print(missoes_por_agencia)
```

```python
# Atividade 2 - Numero máximo de missões feita por uma agencia

missoes_por_agencia = df["Operator"].value_counts()

maior_numero = missoes_por_agencia.max()
agencia = missoes_por_agencia.idxmax()

print("Agência com mais missões:", agencia)
print("Número máximo de missões:", maior_numero)
```
